```python
import redivis
import pandas as pd
```

```python
organization = redivis.organization("AIMI")
dataset = organization.dataset("chexpert_plus")

tables = dataset.list_tables()

for table in tables:
    print(table.properties)
```

```python
main_table = dataset.table("df_chexpert_plus_240401")

variables = main_table.list_variables()

for variable in variables:
    print(variable.properties["name"])
```

```python
df_preview = dataset.query("""
  SELECT
    path_to_image,
    path_to_dcm,
    frontal_lateral,
    ap_pa,
    deid_patient_id,
    patient_report_date_order,
    age,
    sex,
    race,
    ethnicity,
    split,
    section_findings,
    section_impression
  FROM `df_chexpert_plus_240401`
  LIMIT 10
""").to_pandas_dataframe()

df = pd.DataFrame(df_preview)
df.to_csv("explore_data.csv", index=False)
```
